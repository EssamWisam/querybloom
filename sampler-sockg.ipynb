```python
import sys
import os

# Add src directory to path to import kg_voyager module
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..', 'src')))
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..')))  # Add project root for ABSTAT package

from ABSTAT.subgraph_miner import SubgraphMiner
import networkx as nx
import matplotlib.pyplot as plt
```

```python
# Initialize with endpoint and minimum support
miner = SubgraphMiner(
    endpoint='sockg-local',
    min_support=5  # Patterns must appear at least 10 times
)

# Run ABSTAT preprocessing
miner.preprocess(
    property_limit=300,
    max_triples=30000,
    minimize_patterns=True,
    exclude_properties=[
        'label', 'description', 'identifier',   # rdfs/schema noise
    ],
    rand_offset=True
)

print(f"\nPattern pool built with {len(miner.patterns)} frequent patterns")
```

```python
# With max_subgraphs_per_seed=1, the hard upper bound is:
print(f"Upper bound with max_subgraphs_per_seed=1: {len(miner.patterns)}")
```

```python
# More compact visualization - reduce nodesep and ranksep
miner.plot(sample=100, rankdir="LR", include_attributes=True, nodesep=0.1, ranksep=0.4)
```

```python
# Mine subgraphs
print("Mining subgraphs with at least 4 types...\n")

subgraphs = miner.mine(
    # k={3: 0.1, 4: 0.4, 5: 0.3, 6: 0.2},
    k={3: 0.2, 4: 0.4, 5: 0.2, 6: 0.2},
    num_subgraphs=120,
    max_subgraphs_per_seed=1,
    exact_support=False,
    verif_timeout=10,
    m=10,
    output_dir="./subgraphs_output",
    num_attrs=(8, 12),
    uniform_types=True
)

print(f"\nFound {len(subgraphs)} subgraphs!")
```
