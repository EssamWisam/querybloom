```python
import sys
import os
```

```python

# Add current directory to path so we can import abstat.py
current_dir = os.path.abspath('')
if current_dir not in sys.path:
    sys.path.append(current_dir)

from abstat import ABSTAT

# Initialize ABSTAT for SOCKG
abstat = ABSTAT(endpoint="sockg")

# 1. Load Terminology (Subtype Graph)
# Using a limits for testing purposes
abstat.load_terminology(limit=1000)

# 2. Extract Patterns
# Fetch more triples for better coverage
patterns = abstat.extract_patterns(sample_limit=10000)

# Display some patterns
print(f"\nSample Patterns ({len(patterns)} found):")
sorted_patterns = sorted(list(patterns), key=lambda x: (x[1], x[0], x[2]))
for p in sorted_patterns[:30]:
    print(p)
```

```python
# Visualize the schema graph, excluding 'label' properties
abstat.plot_schema_graph(figsize=(16, 12), exclude_properties=['label', 'prefLabel', 'comment'])
```
