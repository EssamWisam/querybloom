# SPARQL Query Executor

Use this notebook to execute SPARQL queries against an endpoint.

```python
import sys
import os
import pandas as pd

# Ensure we can import from the parent directory if not installed as a package
try:
    from kg_voyager.executor.sparql_executor import execute_sparql_query
except ImportError:
    # Append parent directory to path
    sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..')))
    from kg_voyager.executor.sparql_executor import execute_sparql_query
```

```python
# Configuration
ENDPOINT = "sockg-local"  # Replace with 'wikidata' or a specific URL if needed
```

```python
query = """
PREFIX kwg: <http://stko-kwg.geog.ucsb.edu/lod/ontology/>
PREFIX geo: <http://www.opengis.net/ont/geosparql#>
PREFIX sockg: <https://idir.uta.edu/sockg-ontology#>
PREFIX rdfs: <http://www.w3.org/2000/01/rdf-schema#>
PREFIX rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#>

SELECT DISTINCT ?country ?countryLabel ?adm2 ?adm2Label
WHERE {
  ?country a kwg:AdministrativeRegion_0 .
  ?adm2 a kwg:AdministrativeRegion_2 ;
        kwg:sfWithin ?country .
  FILTER EXISTS {
    ?country kwg:sfContains ?loc .
    ?loc a sockg:Location ;
         geo:hasGeometry ?geom .
  }
  OPTIONAL { ?country rdfs:label ?countryLabel }
  OPTIONAL { ?adm2 rdfs:label ?adm2Label }
}

"""

print(f"Executing query on endpoint: {ENDPOINT}...")
result = execute_sparql_query(query, endpoint=ENDPOINT, format_type="csv")

if result.get("success", False):
    df = result.get("results")
    if isinstance(df, pd.DataFrame):
        print(f"Success! Retrieved {len(df)} rows.")
        display(df)
    else:
        print("Success, but result is not a DataFrame:")
        print(df)
else:
    print("Error retrieving results:")
    print(result.get("error", "Unknown error"))
```
